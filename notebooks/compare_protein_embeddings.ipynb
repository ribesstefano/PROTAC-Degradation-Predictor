```python
import torch
from typing import List, Union, Literal
from transformers import AutoTokenizer, AutoModel

def encode_sequence(
    sequence: str,
    tokenizer: AutoTokenizer,
    model: AutoModel,
    device: Union[int, str] = 'cpu',
    pooling: Literal['cls', 'mean', 'sum', 'max', 'mean_sqrt_len', 'none'] = 'none',
) -> torch.Tensor:
    model.to(device)
    model.eval()
    inputs = tokenizer(sequence, return_tensors='pt', padding=True).to(device)
    with torch.no_grad():
        outputs = model(**inputs)
        if pooling == 'cls':
            return outputs.last_hidden_state[:, 0, :]
        elif pooling == 'mean':
            return outputs.last_hidden_state.mean(dim=1)
        elif pooling == 'sum':
            return outputs.last_hidden_state.sum(dim=1)
        elif pooling == 'max':
            return outputs.last_hidden_state.max(dim=1).values
        elif pooling == 'mean_sqrt_len':
            lengths = inputs['attention_mask'].sum(dim=1, keepdim=True).clamp(min=1)
            return outputs.last_hidden_state.sum(dim=1) / lengths.sqrt()
        elif pooling == 'none':
            return outputs.last_hidden_state
        else:
            raise ValueError("Invalid pooling method. Choose from 'cls', 'mean', 'sum', 'max', or 'mean_sqrt_len'.")

def encode_sequences(
        sequences: List[str],
        tokenizer: AutoTokenizer,
        model: AutoModel,
        batch_size: int = 64,
        device: Union[int, str] = 'cpu',
        pooling: Literal['cls', 'mean', 'sum', 'max', 'mean_sqrt_len', 'none'] = 'mean',
) -> torch.Tensor:
    model.to(device)
    model.eval()
    encoded_seqs = []
    with torch.no_grad():
        for i in range(0, len(sequences), batch_size):
            batch = sequences[i:i + batch_size]
            inputs = tokenizer(batch, return_tensors='pt', padding=True).to(device)
            outputs = model(**inputs)
            if pooling == 'cls':
                encoded_seqs.append(outputs.last_hidden_state[:, 0, :])
            elif pooling == 'mean':
                encoded_seqs.append(outputs.last_hidden_state.mean(dim=1))
            elif pooling == 'sum':
                encoded_seqs.append(outputs.last_hidden_state.sum(dim=1))
            elif pooling == 'max':
                encoded_seqs.append(outputs.last_hidden_state.max(dim=1).values)
            elif pooling == 'mean_sqrt_len':
                lengths = inputs['attention_mask'].sum(dim=1, keepdim=True).clamp(min=1)
                encoded_seqs.append(outputs.last_hidden_state.sum(dim=1) / lengths.sqrt())
            else:
                raise ValueError("Invalid pooling method. Choose from 'cls', 'mean', 'sum', 'max', or 'mean_sqrt_len'.")
    return torch.cat(encoded_seqs, dim=0)

poi_seqs = df["POI Sequence"].unique().tolist()[:5]
emb = encode_sequences(
    poi_seqs,
    tokenizer=tokenizer,
    model=model,
    batch_size=64,
    device='cpu',
    pooling='mean'
)
emb.shape
```

```python
import torch
import torch.nn.functional as F
from sklearn.metrics.pairwise import cosine_similarity

# ----------------------------------------------------------
# Assume:
#   EA : torch.Tensor, shape (1, La, D)  – residues of protein A
#   EB : torch.Tensor, shape (1, Lb, D)  – residues of protein B
# Output:
#   z  : torch.Tensor, shape (1, D)      – co-pooled embedding
# ----------------------------------------------------------

# ------------------------------------------------------------------
# 1. Similarity-Weighted (SW) pooling
# ------------------------------------------------------------------
def sw_pool(EA: torch.Tensor, EB: torch.Tensor) -> torch.Tensor:
    """
    Parameter-free co-pooling.
    Steps (batch dim=1 omitted for clarity):
      1. Scaled cosine-similarity matrix S  ∈ ℝ^{La×Lb}
      2. Global softmax → attention weights over all residue pairs
      3. Marginalise to get α (La) and β (Lb)
      4. Weighted sums → vA, vB  ∈ ℝ^{D}
      5. Return their mean                       (1, D)
    """
    B, La, D = EA.shape
    _, Lb, _ = EB.shape

    # (1, La, D) × (1, D, Lb) → (1, La, Lb)
    S = torch.matmul(EA, EB.transpose(1, 2)) / D**0.5        # similarity matrix
    S = S.view(-1, La * Lb)                                  # (1, La*Lb)

    A = torch.softmax(S, dim=-1)                             # softmax over all pairs
    A = A.view(La, Lb)                                       # (La, Lb)

    alpha = A.sum(dim=1)                                     # (La)
    beta  = A.sum(dim=0)                                     # (Lb)

    # Weighted residue sums
    vA = (alpha[:, None] * EA[0]).sum(dim=0)                 # (D)
    vB = (beta[:, None] * EB[0]).sum(dim=0)                  # (D)

    z = 0.5 * (vA + vB)                                      # (D)
    return z.unsqueeze(0)                                    # (1, D)

# ------------------------------------------------------------------
# 2. Optimal-Transport (OT) pooling  (entropic Sinkhorn, ε = 0.1)
# ------------------------------------------------------------------
def sinkhorn_transport(cost: torch.Tensor,
                       iters: int = 20,
                       eps: float = 0.1) -> torch.Tensor:
    """
    cost : (La, Lb) – pairwise costs   (here: 1 - cosine)
    Returns doubly-stochastic plan P  (La, Lb)
    """
    La, Lb = cost.shape
    # K = exp(-cost / eps)
    K = torch.exp(-cost / eps)

    u = torch.ones(La, 1, device=cost.device) / La
    v = torch.ones(Lb, 1, device=cost.device) / Lb

    K_t = K.transpose(0, 1)

    for _ in range(iters):
        u = 1.0 / (K @ v)
        v = 1.0 / (K_t @ u)

    P = torch.diagflat(u.squeeze()) @ K @ torch.diagflat(v.squeeze())  # (La, Lb)
    return P

def ot_pool(EA: torch.Tensor, EB: torch.Tensor,
            eps: float = 0.1, iters: int = 20) -> torch.Tensor:
    """
    OT pooling:
      1. Cost = 1 - cosine(EA_i, EB_j)
      2. Sinkhorn → transport plan P
      3. Pool with element-wise product kernel ⇒ Σ_{ij} P_ij (EA_i ⊙ EB_j)
    Output: (1, D)
    """
    La, D = EA.shape[1:]
    Lb     = EB.shape[1]

    # --- pairwise cosine similarity ---------------------------------
    # (La, D) ↔ (Lb, D)  →  (La, Lb)
    EA_norm = F.normalize(EA[0], dim=-1)                       # (La, D)
    EB_norm = F.normalize(EB[0], dim=-1)                       # (Lb, D)
    cos_sim = torch.matmul(EA_norm, EB_norm.T)                 # (La, Lb)
    cost    = 1.0 - cos_sim                                    # (La, Lb)

    # --- Sinkhorn plan ---------------------------------------------
    P = sinkhorn_transport(cost, iters=iters, eps=eps)         # (La, Lb)

    # --- OT pooling kernel  φ(x,y) = x ⊙ y --------------------------
    # (La, Lb, D): broadcasted outer product
    kernel = EA[0][:, None, :] * EB[0][None, :, :]             # (La, Lb, D)
    z = (P.unsqueeze(-1) * kernel).sum(dim=(0, 1))             # (D)
    return z.unsqueeze(0)                                      # (1, D)


poi_mean_emb = encode_sequence(poi, tokenizer=tokenizer, model=model, pooling='mean').cpu().numpy()
e3_mean_emb = encode_sequence(e3, tokenizer=tokenizer, model=model, pooling='mean').cpu().numpy()

poi_emb = encode_sequence(poi, tokenizer=tokenizer, model=model)
e3_emb = encode_sequence(e3, tokenizer=tokenizer, model=model)

sw_emb = sw_pool(poi_emb, e3_emb).cpu().numpy()
print(f"SW-pool embedding shape: {emb.shape}")

ot_emb = ot_pool(poi_emb, e3_emb).cpu().numpy()
print(f"OT-pool embedding shape: {emb.shape}")

# Print cosine similarity between POI and E3 embeddings
cos_sim = cosine_similarity(sw_emb, ot_emb)
print(f"Cosine similarity between SW and OT embeddings: {cos_sim[0][0]:.4f}")

# Print cosine similarity between POI and E3 embeddings
cos_sim = cosine_similarity(poi_mean_emb.reshape(1, -1), e3_mean_emb.reshape(1, -1))
print(f"Cosine similarity between POI and E3 embeddings: {cos_sim[0][0]:.4f}")

# Print cosine similarity between POI and SW embeddings
cos_sim = cosine_similarity(poi_mean_emb.reshape(1, -1), sw_emb.reshape(1, -1))
print(f"Cosine similarity between POI and SW embeddings: {cos_sim[0][0]:.4f}")
# Print cosine similarity between E3 and SW embeddings
cos_sim = cosine_similarity(e3_mean_emb.reshape(1, -1), sw_emb.reshape(1, -1))
print(f"Cosine similarity between E3 and SW embeddings: {cos_sim[0][0]:.4f}")
# Print cosine similarity between POI and OT embeddings
cos_sim = cosine_similarity(poi_mean_emb.reshape(1, -1), ot_emb.reshape(1, -1))
print(f"Cosine similarity between POI and OT embeddings: {cos_sim[0][0]:.4f}")
# Print cosine similarity between E3 and OT embeddings
cos_sim = cosine_similarity(e3_mean_emb.reshape(1, -1), ot_emb.reshape(1, -1))
print(f"Cosine similarity between E3 and OT embeddings: {cos_sim[0][0]:.4f}")
```

```python
from biotite.sequence import ProteinSequence
from biotite.sequence.align import align_optimal, SubstitutionMatrix
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
from tqdm import tqdm

# sequences = df['E3 Ligase Sequence'].dropna().unique().tolist()
sequences = df['POI Sequence'].dropna().unique().tolist()[:10]

sequences = [s.replace('U', 'C') for s in sequences if isinstance(s, str)]

# Create ProteinSequence objects
sequences = [ProteinSequence(seq) for seq in sequences]

# Define necessary variables
num_sequences = len(sequences)

# Initialize the percent identity matrix
percent_identity_matrix = np.eye(num_sequences)

# Use the standard BLOSUM62 substitution matrix
substitution_matrix = SubstitutionMatrix.std_protein_matrix()

def get_identity_percentage(
        seq1: ProteinSequence,
        seq2: ProteinSequence,
        substitution_matrix: SubstitutionMatrix,
) -> float:
    """ Calculate the percent identity between two sequences using global alignment.
    
    Args:
        seq1 (ProteinSequence): The first protein sequence.
        seq2 (ProteinSequence): The second protein sequence.
        substitution_matrix (SubstitutionMatrix): The substitution matrix to use for alignment.
        
    Returns:
        float: The percent identity between the two sequences.
    """
    # Perform global alignment
    alignments = align_optimal(seq1, seq2, substitution_matrix, local=False)
    alignment = alignments[0]
    aligned_seq1, aligned_seq2 = alignment.get_gapped_sequences()

    # Count the number of matching characters
    matched_chars = sum(a == b for a, b in zip(aligned_seq1, aligned_seq2) if a != '-' and b != '-')
    total_length = max(len(seq1), len(seq2))

    # Calculate percent identity
    identity_percentage = (matched_chars / total_length)
    return identity_percentage

# Compute the identity matrix
for i in tqdm(range(num_sequences), desc="Computing percent identity matrix"):
    for j in range(i + 1, num_sequences):
        # Calculate percent identity
        identity_percentage = get_identity_percentage(
            sequences[i],
            sequences[j],
            substitution_matrix,
        )
        percent_identity_matrix[i, j] = identity_percentage
        percent_identity_matrix[j, i] = identity_percentage
print("Percent identity matrix computed.")

# Create a DataFrame
percent_identity_df = pd.DataFrame(percent_identity_matrix)

# Plot a heatmap without axis labels
annot = True if len(sequences) < 20 else False  # Annotate only for small matrices
plt.figure(figsize=(10, 8))
sns.heatmap(percent_identity_df, cmap="viridis", annot=annot, fmt=".1%", cbar_kws={'label': 'Percent Identity'})
plt.title("Percent Identity Matrix Heatmap")
# Here we explicitly hide the x and y axis labels for a cleaner figure
plt.xlabel("")
plt.ylabel("")
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
from Bio.Align import PairwiseAligner, substitution_matrices
import numpy as np

# ----------------------------------------------------------------------
# 1.  PREPARATION
# ----------------------------------------------------------------------
# list[ProteinSequence]
# sequences = df['E3 Ligase Sequence'].dropna().unique().tolist()
sequences = df['POI Sequence'].dropna().unique().tolist()[:10]
num_sequences = len(sequences)

# BLOSUM62 + commonly used affine gap penalties
subs_mat = substitution_matrices.load("BLOSUM62")
aligner = PairwiseAligner()
aligner.substitution_matrix = subs_mat
aligner.mode               = "global"          # Needleman–Wunsch
aligner.open_gap_score     = -10               # standard BLASTP
aligner.extend_gap_score   = -0.5              #  … idem

# ----------------------------------------------------------------------
# 2.  SELF-SCORES  (needed only once)
# ----------------------------------------------------------------------
self_scores = np.zeros(num_sequences)
for i, seq in enumerate(sequences):
    self_scores[i] = aligner.score(seq, seq)   # S_ii

# ----------------------------------------------------------------------
# 3.  PAIRWISE  NORMALISED  SCORES
# ----------------------------------------------------------------------
nas_matrix = np.eye(num_sequences, dtype=float)  # NAS_ii = 1.0

for i in range(num_sequences):
    for j in range(i + 1, num_sequences):
        raw_score = aligner.score(sequences[i], sequences[j])      # S_ij
        denom = (self_scores[i] * self_scores[j]) ** 0.5           # √(S_ii S_jj)
        nas = max(raw_score / denom, 0)                            # bound to [0,1]
        nas_matrix[i, j] = nas_matrix[j, i] = nas

# ----------------------------------------------------------------------
# 4.  OPTIONAL:  export  identity  &  positives  for reporting
# ----------------------------------------------------------------------
def id_pos(aln):
    s1, s2 = aln.aligned[0], aln.aligned[1]    # coordinate pairs
    matches = positives = 0
    for (a0, a1), (b0, b1) in zip(s1, s2):
        for k in range(a1 - a0):
            aa, bb = aln.sequence[0][a0 + k], aln.sequence[1][b0 + k]
            if aa == bb:
                matches += 1
            if subs_mat[(aa, bb)] > 0:
                positives += 1
    length = sum(a1 - a0 for (a0, a1) in s1)
    return matches / length, positives / length

# Create a DataFrame
nas_df = pd.DataFrame(nas_matrix)

# Plot a heatmap without axis labels
annot = True if len(sequences) < 20 else False  # Annotate only for small matrices
plt.figure(figsize=(10, 8))
sns.heatmap(nas_df, cmap="viridis", annot=annot, fmt=".2f", cbar_kws={'label': 'NAS (Normalised Global Alignment Score)'})
plt.title("Normalised Global Alignment Score Matrix Heatmap")
# Here we explicitly hide the x and y axis labels for a cleaner figure
plt.xlabel("")
plt.ylabel("")
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
import os
import yaml
from typing import Optional

# Wrapper class for forcing single quotes on SMILES only
class QuotedStr(str):
    pass

def quoted_str_representer(dumper, data):
    # Force single-quoted scalar style for QuotedStr objects only
    return dumper.represent_scalar('tag:yaml.org,2002:str', data, style="'")

yaml.add_representer(QuotedStr, quoted_str_representer)

# Forcing flow style for id lists
class FlowList(list):
    pass

def flow_list_representer(dumper, data):
    return dumper.represent_sequence('tag:yaml.org,2002:seq', data, flow_style=True)

yaml.add_representer(FlowList, flow_list_representer)

def write_boltz2_yaml(
        filename: str,
        poi_sequence: Optional[str] = None,
        e3_sequence: Optional[str] = None,
        smiles: Optional[str] = None,
):
    """
    Write POI and E3 ligase sequences and SMILES to a YAML file in Boltz-2 format.

    Args:
        filename (str): Path to the output YAML file.
        poi_sequence (str): Protein of Interest (POI) sequence.
        e3_sequence (str): E3 ligase sequence.
        smiles (str): SMILES string.
    """
    if not any([poi_sequence, e3_sequence]):
        raise ValueError("At least one parameter (poi_sequence, e3_sequence, smiles) must be provided.")

    sequences = []
    if poi_sequence is not None:
        sequences.append({"protein": {
            "id": FlowList(["A"]),
            "sequence": poi_sequence,
        }})
    if e3_sequence is not None:
        sequences.append({"protein": {
            "id": FlowList(["B"]),
            "sequence": e3_sequence,
        }})
    if smiles is not None:
        sequences.append({"ligand": {
            "id": FlowList(["C"]),
            "smiles": QuotedStr(smiles),
        }})
    
    boltz2_dict = {
        "sequences": sequences,
    }

    with open(filename, "w") as f:
        yaml.dump(boltz2_dict, f, sort_keys=False)

# Example usage:
boltz_dir = "../data/boltz2"
os.makedirs(boltz_dir, exist_ok=True)
write_boltz2_yaml(
    f"{boltz_dir}/test.yaml",
    selected_row["POI Sequence"].values[0],
    selected_row["E3 Ligase Sequence"].values[0],
    selected_row["Smiles"].values[0],
)

# Show the plain content of the YAML file
with open(f"{boltz_dir}/test.yaml", "r") as f:
    print(f.read())
```

```python
# Create a hashed ID for the DataFrame that includes the columns: "Smiles", "POI Sequence", "E3 Ligase Sequence"
import hashlib

def get_hash(s: str) -> str:
    """
    Generate a SHA-256 hash for a given string.
    
    Args:
        s (str): The input string to hash.
        
    Returns:
        str: The SHA-256 hash of the input string.
    """
    return hashlib.sha256(s.encode()).hexdigest()

def create_hashed_id(row):
    """
    Create a hashed ID for a row based on the 'Smiles', 'POI Sequence', and 'E3 Ligase Sequence' columns.
    
    Args:
        row (pd.Series): A row from the DataFrame.
        
    Returns:
        str: A hashed ID.
    """
    hash_input = f"{row['Smiles']}_{row['POI Sequence']}_{row['E3 Ligase Sequence']}"
    return get_hash(hash_input)

# Apply the function to create a new column 'Hashed ID'
df['Hashed ID'] = df.apply(create_hashed_id, axis=1)
# Show the DataFrame with the new 'Hashed ID' column
print(df[['Hashed ID']].nunique())
print(len(df))
```

```python
from tqdm import tqdm

boltz_dir = "../data/boltz2"
os.makedirs(boltz_dir, exist_ok=True)

for _, row in tqdm(df.iterrows(), desc="Writing Complexes Boltz2 YAML files", total=len(df)):
    # Skip if any of the sequences are NaN
    if pd.isna(row["POI Sequence"]) or pd.isna(row["E3 Ligase Sequence"]) or pd.isna(row["Smiles"]):
        continue
    filename = f"{boltz_dir}/complex_{row['Hashed ID']}.yaml"
    write_boltz2_yaml(
        filename,
        row["POI Sequence"],
        row["E3 Ligase Sequence"],
        row["Smiles"],
    )

# Loop through all the unique POI sequences and write them to YAML files
for _, row in tqdm(df.iterrows(), desc="Writing POI Boltz2 YAML files", total=len(df)):
    # Skip if any of the sequences are NaN
    if pd.isna(row["POI Sequence"]) or pd.isna(row["E3 Ligase Sequence"]) or pd.isna(row["Smiles"]):
        continue
    filename = f"{boltz_dir}/poi_{get_hash(row['POI Sequence'])}.yaml"
    if os.path.exists(filename):
        continue
    # Write the POI sequence to a YAML file
    write_boltz2_yaml(
        filename,
        row["POI Sequence"],
    )

# Loop through all the unique E3 ligase sequences and write them to YAML files
for _, row in tqdm(df.iterrows(), desc="Writing E3 ligase Boltz2 YAML files", total=len(df)):
    # Skip if any of the sequences are NaN
    if pd.isna(row["POI Sequence"]) or pd.isna(row["E3 Ligase Sequence"]) or pd.isna(row["Smiles"]):
        continue
    filename = f"{boltz_dir}/e3l_{get_hash(row['E3 Ligase Sequence'])}.yaml"
    if os.path.exists(filename):
        continue
    # Write the E3 ligase sequence to a YAML file
    write_boltz2_yaml(
        filename,
        e3_sequence=row["E3 Ligase Sequence"],
    )
```
